```python
data = [
 {
 "customer": " Alice ",
 "city": "Delhi",
 "orders": [
 {"product": "Laptop", "price": "70000", "quantity": 1},
 {"product": "Mouse", "price": "1200", "quantity": 2}
 ]
 },
 {
 "customer": "BOB",
 "city": "Mumbai",
 "orders": [
 {"product": "Monitor", "price": "15000", "quantity": 2},
 {"product": "Keyboard", "price": "2500", "quantity": 1}
 ]
 },
 {
 "customer": " Charlie ",
 "city": "Delhi",
 "orders": [
 {"product": "Laptop", "price": "70000", "quantity": 1},
 {"product": "Monitor", "price": "15000", "quantity": 1}
 ]
 }
]
```

```python
# 1. Clean customer names. 
for i in data:
    print(i['customer'].strip())


```

```python
# 2. Convert prices and quantities to numeric types. 
for i in data:
    res = i['orders']
    for j in res:
        j['price'] = int(j['price'])            #str to int 
    print(type(j['price']))
```

```python
# 3. Calculate every item's total. 
for i in data:
    res = i['orders']
    for j in res:
        j['total']=j['price']*j['quantity']
        print(j)

    
```

```python
# 4.  Calculate every customer's total spending.  
totalr = {}
flattened = []
for i in data:
    customers =i['customer'].lstrip().upper()
    res = i['orders'] 
    total_spending = 0
    net_price = 0
    for j in res:
        net_amount =  j['price'] * j ['quantity']
        total_spending = total_spending + j['total'] 
        flattened.append({'customer':customers , 'city':i['city'] , 'product':j['product'] , 'price': j['price'],'quantity':j['quantity'],'net_amount':net_amount,'total_spending':total_spending})
    totalr[customers] = total_spending
print(flattened)
import pandas as pd
df = pd.DataFrame(flattened)
df.display()
print(totalr)
        
        
```

```python
# 8. Calculate total revenue.
total_revenue = 0
for  i in data:
    res = i['orders']
    for j in res:
        total_revenue = total_revenue + j['total']
print(total_revenue)

```

```python
# 9. Calculate revenue by city. 
mumbai = 0
delhi = 0
for i in data:
    city=i['city']
    orders = i['orders']

    for j in orders:
        total = 0
        if city=='Delhi':
            total = j['price']*j['quantity']
            delhi = delhi + total
        elif city == "Mumbai":
            total = j['price']*j['quantity']
            mumbai = mumbai + total
print("Mumbai" , mumbai)

print("Delhi" , delhi)


```

```python
# 10. Calculate revenue by product.
# it is use where you clean your data and add new columns such as  total   and some other in data 
product = {}
for i in data:
    res = i['orders']
    for j in res:
        if j['product'] not in product:
            product[j['product']] = j['total']
        else:
            product[j['product']] = product[j['product']] + j['total']
print(product)

```

```python
# Q2.
text = """
Python is powerful.
Python is easy to learn.
Learning Python is fun.
Python is widely used.
"""
```

```python
# 1. Converts the text to lowercase. 
print(text.lower())
```

```python
# 2. Removes punctuation. 
print(text.replace(".",""))
```

```python
# 3. Counts each word.
x = text.replace(".","") 
for i in x.split():
    print(i , x.count(i))
```

```python
print(x)
l = []
for i in x.split():
    l.append(i)
print(l)

```

```python
for j in l:
    print( j   , len(j))
```

```python
# 4. Finds the most frequently occurring word. 
x = text.replace(".","") 
max = 0

word = ""
for i in x.split():
    if x.count(i)> max:
        max = x.count(i)
        word = i
print(word , max)
    
```

```python
# Prints the total number of words and unique words.
words = x.split()
total_words = len(words)
unique_words = len(set(words))
print(f"Total words: {total_words}")
print(f"Unique words: {unique_words}")

```
