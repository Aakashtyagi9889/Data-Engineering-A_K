# Python Programming Language

## Overview
Python is a high-level, interpreted, and general-purpose programming language created by Guido van Rossum and first released in 1991. It emphasizes code readability with its notable use of significant whitespace.

## Key Features
* **Easy to Learn**: Simple, clean syntax that resembles natural language
* **Abstraction level**: High level Language 
* **Interpreted**: Code is executed line by line, making debugging easier
* **Dynamically Typed**: No need to declare variable types explicitly
* **Multi-paradigm**: Supports procedural, object-oriented, and functional programming
* **Extensive Libraries**: Rich ecosystem with packages for data science, web development, automation, and more
* **Cross-platform**: Runs on Windows, macOS, Linux, and other operating systems

## Popular Use Cases
* Data Science & Machine Learning (pandas, NumPy, scikit-learn, TensorFlow)
* Web Development (Django, Flask, FastAPI)
* Automation & Scripting
* Data Engineering (PySpark, Apache Airflow)
* Scientific Computing
* API Development

```python
#Data Types in python -->String , list , tuples , set , dict 
# string =  It store alphanumeric text
# Features - Mutability -------->Immutable
a = "I Love DE 101"
type(a)

print(a[3]) #INDEXING
print(a[-6]) # REVERSE INDEXING
print(a[7:9]) # Slicing 
print(a[0:13:2]) 
print(a[-3:]) # Reverse Slicing 
print(a[-6:-4])

# Reverse the string  (Interview Question)
print(a[::-1])

# Love ==> evoL
print(a[-8:-12  :-1])
# without using negative slicing
print(a[2:6][::-1])
```
