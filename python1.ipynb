# Python Programming Language

## Overview
Python is a high-level, interpreted, and general-purpose programming language created by Guido van Rossum and first released in 1991. It emphasizes code readability with its notable use of significant whitespace.

## Key Features
* **Easy to Learn**: Simple, clean syntax that resembles natural language
* **Abstraction level**: High level Language 
* **Interpreted**: Code is executed line by line, making debugging easier
* **Dynamically Typed**: No need to declare variable types explicitly
* **Multi-paradigm**: Supports procedural, object-oriented, and functional programming
* **Extensive Libraries**: Rich ecosystem with packages for data science, web development, automation, and more
* **Cross-platform**: Runs on Windows, macOS, Linux, and other operating systems

## Popular Use Cases
* Data Science & Machine Learning (pandas, NumPy, scikit-learn, TensorFlow)
* Web Development (Django, Flask, FastAPI)
* Automation & Scripting
* Data Engineering (PySpark, Apache Airflow)
* Scientific Computing
* API Development

```python
# functions
def count_5(st,x,delimiter):
    st = st.replace(delimiter,"")
    st_list = st.split()
    L = []
    for i in st_list:
        if len(i) == x:
          L.append(i)

    print(len(L))
```

```python
# Different methods and modes for reading CSV in pandas

import pandas as pd

# 1. Basic read_csv
df = pd.read_csv('file.csv')

# 2. Common parameters:

# Delimiter/separator
df = pd.read_csv('file.csv', sep=',')  # default comma
df = pd.read_csv('file.tsv', sep='\t')  # tab-separated
df = pd.read_csv('file.txt', delimiter='|')  # pipe-separated

# Header options
df = pd.read_csv('file.csv', header=0)  # first row as column names (default)
df = pd.read_csv('file.csv', header=None)  # no header row
df = pd.read_csv('file.csv', names=['col1', 'col2', 'col3'])  # specify column names

# Index column
df = pd.read_csv('file.csv', index_col=0)  # use first column as index
df = pd.read_csv('file.csv', index_col='id')  # use specific column as index

# Data types
df = pd.read_csv('file.csv', dtype={'col1': str, 'col2': int})
df = pd.read_csv('file.csv', dtype=str)  # all columns as strings

# Handle missing values
df = pd.read_csv('file.csv', na_values=['NA', 'null', ''])  # specify NA values
df = pd.read_csv('file.csv', keep_default_na=False)  # don't convert to NaN

# Skip rows
df = pd.read_csv('file.csv', skiprows=5)  # skip first 5 rows
df = pd.read_csv('file.csv', skiprows=[0, 2, 4])  # skip specific rows
df = pd.read_csv('file.csv', skipfooter=3)  # skip last 3 rows

# Read specific columns
df = pd.read_csv('file.csv', usecols=['col1', 'col3'])  # by name
df = pd.read_csv('file.csv', usecols=[0, 2, 4])  # by index

# Parse dates
df = pd.read_csv('file.csv', parse_dates=['date_column'])
df = pd.read_csv('file.csv', parse_dates={'datetime': ['date', 'time']})

# Encoding
df = pd.read_csv('file.csv', encoding='utf-8')  # default
df = pd.read_csv('file.csv', encoding='latin-1')


### utf-8 is the default encoding for pandas.read_csv. If your file is encoded in a different format, you can specify the encoding using the encoding parameter.
##special Char. 

## latin-1


## Dealing with Excel - cp1252




# Chunking (for large files)
chunk_iter = pd.read_csv('large_file.csv', chunksize=10000)
for chunk in chunk_iter:
    print(chunk.shape)  # process each chunk

# Number of rows
df = pd.read_csv('file.csv', nrows=100)  # read only first 100 rows

# Compression
df = pd.read_csv('file.csv.gz', compression='gzip')
df = pd.read_csv('file.csv.zip', compression='zip')
df = pd.read_csv('file.csv', compression='infer')  # auto-detect (default)

# Quoting
df = pd.read_csv('file.csv', quotechar='"')  # default
df = pd.read_csv('file.csv', quoting=0)  # QUOTE_MINIMAL

# Low memory mode
df = pd.read_csv('file.csv', low_memory=False)  # read entire file at once

# Error handling
df = pd.read_csv('file.csv', error_bad_lines=False)  # skip bad lines (deprecated)
df = pd.read_csv('file.csv', on_bad_lines='skip')  # skip bad lines (new)
df = pd.read_csv('file.csv', on_bad_lines='warn')  # warn on bad lines

# Engine
df = pd.read_csv('file.csv', engine='c')  # C engine (faster, default)
df = pd.read_csv('file.csv', engine='python')  # Python engine (more features)

# Converters
df = pd.read_csv('file.csv', converters={'col1': lambda x: x.upper()})

# Comments
df = pd.read_csv('file.csv', comment='#')  # skip lines starting with #

print("CSV reading methods and modes demonstrated!")
```
